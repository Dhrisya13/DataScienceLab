```python
import pandas as pd
import matplotlib.pyplot as plt
a=[4,5,6]
var=pd.Series(a)
print(var)
print('0th index =',var[0])
```

```python
var=pd.Series(a,index=['X','Y','Z'])
print(var)
print('Y value = ', var['Y'])
```

```python
calories = {"day1":150,"day2":200,"day3":250}
var=pd.Series(calories)
print(var)
```

```python
data={"calories":[150,200,250],"duration":[50,40,60]}
var=pd.DataFrame(data)
print(var)
```
