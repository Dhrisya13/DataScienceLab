```python
#piechart
y=np.array([25,65,30,75])
plt.pie(y)
plt.show()
```

```python
y=np.array([25,65,30,75])
ylabels=["Melon","Orange","Mango","Apple"]
plt.pie(y,labels=ylabels)
plt.show()
```

```python
y=np.array([25,65,30,75])
ylabels=["Melon","Orange","Mango","Apple"]
yexplode=[0,0.2,0,0]
plt.pie(y,labels=ylabels,explode=yexplode)
plt.show()
```

```python
y=np.array([25,65,30,75])
ylabels=["Melon","Orange","Mango","Apple"]
yexplode=[0,0.2,0,0]
plt.pie(y,labels=ylabels,explode=yexplode, shadow=True)
plt.show()
```

```python
y=np.array([25,65,30,75])
ylabels=["Melon","Orange","Mango","Apple"]
ycolors=["red","green","gold","blue"]
plt.pie(y,labels=ylabels,colors=ycolors)
plt.show()
```

```python
y=np.array([25,65,30,75])
ylabels=["Melon","Orange","Mango","Apple"]
plt.pie(y,labels=ylabels)
plt.legend(title='fruits')
plt.show()
```
