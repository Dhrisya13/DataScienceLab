```python
import numpy as np
import matplotlib.pyplot as plt
xpoints=np.array([0,6])
ypoints=np.array([0,10])

plt.plot(xpoints,ypoints)
plt.show()
```

```python
ypoints=np.array([0,5,6,4,2])
plt.plot(ypoints, marker = 'o')
plt.show()
```

```python
plt.plot(ypoints,marker='*')
plt.show()
```

```python
plt.plot(ypoints,marker='h')
plt.plot(ypoints, 'o:r', ms='20',mec='g')
plt.show()
```

```python
ypoints=np.array([3,5,1,9])
plt.plot(ypoints, linestyle='dotted')
plt.show()
```

```python
plt.plot(ypoints,linestyle='dashed')
plt.show()
```

```python
plt.plot(ypoints, ls='-.')
plt.show()
```

```python
plt.plot(ypoints, color='black')
plt.show()
```

```python
plt.plot(ypoints, linewidth='6')
plt.show()
```

```python
#labels
x=np.array([25,30,10,20,25,50,35,40])
y=np.array([100,200,300,400,500,600,700,800])
plt.plot(x,y)

plt.xlabel('Average weight')
plt.ylabel('Measures')
plt.show()
```

```python
x=np.array([20,30,25,30,45,55,40,50])
y=np.array([100,200,300,400,500,600,700,800])
plt.plot(x,y)
plt.title('Weight Data')
plt.xlabel('Average weight')
plt.ylabel('Measures')
plt.show()
```

```python
plt.plot(x,y)
font1={'color':'green','size':'20'}
font2={'color':'red','size':'10'}
plt.title('Weight Data',fontdict=font1)
plt.xlabel('Weight',fontdict=font2)
plt.ylabel('Measures',fontdict=font2)
plt.show()
```

```python
plt.plot(x,y)
plt.title('Weight Data',loc='left')
plt.xlabel('Weight')
plt.ylabel('Measures')
plt.show()
```

```python
plt.plot(x,y)
plt.title('Weight Data')
plt.xlabel('Weight')
plt.ylabel('Measures')
plt.grid()
plt.show()
```

```python
plt.plot(x,y)
plt.title('Weight Data',loc='left')
plt.xlabel('Weight')
plt.ylabel('Measures')
plt.grid(axis='x')
plt.show()
```

```python
plt.plot(x,y)
plt.title('Weight Data',loc='left')
plt.xlabel('Weight')
plt.ylabel('Measures')
plt.grid(axis='y')
plt.show()
```

```python
plt.plot(x,y)
plt.title('Weight Data',loc='left')
plt.xlabel('Weight')
plt.ylabel('Measures')
plt.grid(color='blue', linestyle='--', linewidth=0.5)
plt.show()
```
