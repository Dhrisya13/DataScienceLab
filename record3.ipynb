```python
import numpy as np
import matplotlib.pyplot as plt

x=np.array([1,5])
y=np.array([0,5])
plt.subplot(2,3,1)
plt.title("Sales")
plt.xlabel("Size",color='blue')
plt.ylabel("Measure",color='blue')
plt.scatter(x,y)

a=np.array([0,8,5,15,10])
b=np.array([1,4,8,6,2])
plt.subplot(2,3,2)
plt.plot(a,marker='*',color='Orange',linewidth=2.5)
plt.plot(b,linestyle='--',color='blue')

a=np.array([0,8,5,15,10])
b=np.array([1,4,8,6,2])
plt.subplot(2,3,3)
plt.plot(a,color='blue',marker='o',mfc='yellow',mec='black')

a=np.array([0,8,5,15,10])
b=np.array([1,4,8,6,2])
plt.subplot(2,3,4)
plt.title("Measures")
plt.grid(axis='x')
plt.plot(x,y)

a=np.array([0,8,5,15,10])
b=np.array([1,4,8,6,2])
plt.subplot(2,3,5)
size=np.array([400,500,50,50,300])
color=np.array(['red','blue','green','orange','black'])
plt.scatter(a,b,s=size,c=color,alpha=0.5)

a=np.array([0,8,5,15,10])
b=np.array([1,4,8,6,2])
plt.subplot(2,3,6)
plt.bar(a,b,color='red',width=0.5)
plt.show()
```

```python
y = np.array([35, 25, 20,20])
mylabels=[35,25,20,20]
myexplode = [0.2, 0, 0, 0]

plt.pie(y, explode = myexplode,startangle=90, labels=mylabels, shadow = True)
plt.legend()
plt.show() 
```
